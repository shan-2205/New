# 3.. Tools, Permissions & the Trust Model ( CRITICAL — Don't Skip ! ) 🚨


```python
'''

═══════════════════════════════════════════════
🎯 WHAT ARE TOOLS ?
═══════════════════════════════════════════════


Codex is an AI coding agent that lives inside your terminal ..
            OR
It is an LLM along with , it have access to TOOls to get the work done !

    Examples :

    For Reading files from your workspace                 → Codex uses Read Tool
    For Editing code , creating new file , updating file  → Codex uses Edit Tool
    For Running terminal commands                         → Codex uses Shell Tool
                                                            Example :
                                                                    npm install


🚨 The IMPORTANT thing to understand :

    👉 You NEVER manually pick which tool to use !
    👉 Codex AUTOMATICALLY decides :
            - WHICH tool is needed
            - WHEN to use it
            
            
            
            
            
═══════════════════════════════════════════════
🎯 WHAT ARE PERMISSIONS ?
═══════════════════════════════════════════════


Now — just because Codex CAN use tools …
doesn't mean it should do EVERYTHING without asking you first !

This is where PERMISSIONS come in 🚦


    Permissions = Rules that control WHAT Codex is allowed to do
                  & WHEN it must stop and ask for your approval






There are 2 types of Permissions :
1 ) User-level ( it means , these permissions are applicable for all the projects of your system )

        # ~/.codex/config.toml   ( user-level )
                            OR
            C:\Users\shant\.codex\config.toml
    



2 ) Project-level Permissions : These are stored inside .codex/config.toml file  inside a TRUSTED project

        SALES_ANALYSIS/
          ├── .agents/         
          ├── .codex/           ← CREATE THIS
          │     └── config.toml ← PUT CONFIG HERE
          ├── data/
          ├── reports/
          ├── src/
          ├── tests/
          ├── AGENTS.md
          └── Sales_Analysis-threat-model.md






config.toml have 3 things : 
    I)  File-system permissions
    II) MCP servers Permissions
    III) settings
    
    
    
    


Section               Controls                              Key params
──────────────────────────────────────────────────────────────────────────────
I.  Filesystem        What Codex can touch &                sandbox_mode
    permissions       when it asks                          approval_policy
                                                            network_access

II. MCP servers       Which external tools                  [mcp_servers.name]
                      Codex can connect to                  eg : [mcp_servers.github] # Codex can access your GitHub repos , PRs , Issues
                                                                 [mcp_servers.postgres] # Codex can now query your PostgreSQL database

III. Settings         Model, auth, web search,              model
                      profiles                              cli_auth_credentials_store
                                                            web_search
                                                            profiles
                                                            
                                                            
    [ VS CODE] : show below command
        /permissions  -> view & adjust what Codex is allowed to do
        /status       -> see the CURRENT approval policy & sandbox mode                                                            





# ── I. PERMISSIONS ──────────────────────────────
### How much of your FILESYSTEM Codex can access
approval_policy = "on-request"                          # "on-request" | "untrusted" | "never"
                👉 "on-request" : Codex asks YOU before doing risky / irreversible actions
                👉 "untrusted"  : Codex asks before running ALMOST every command
                                   ( strictest / maximum oversight )
                👉 "never"      : Codex never asks — runs everything automatically
                                   ( for CI/CD pipelines only ⚠️ )

sandbox_mode    = "workspace-write"                     # "read-only" | "workspace-write" | "danger-full-access"
                👉 "workspace-write" = Codex can read + write INSIDE the project folder , ie Cannot touch anything OUTSIDE the project
                👉 "read-only" → Codex can only READ files — no writing, no editing
                                          ( safest mode )
                👉 "danger-full-access" → No sandbox — Codex can touch anything on your machine
                                          
                
                
                
[sandbox_workspace_write]        ## [sandbox_workspace_write] is a TOML section header  — not a parameter itself.  
## 👉 It means: "These settings apply ONLY when sandbox_mode = workspace-write is active"

network_access = false                                  # true | false
                👉 false  → Shell commands don't have internet access
                            ( safe default — blocks accidental data leaks ✅ )
                👉 true   → Shell commands have access of internet
                            ( only enable when your workflow needs it ⚠️ )



# ── II. MCP SERVERS ──────────────────────────────
# which external tools Codex can connect to ( ie  MCP servers listed here are allowed to run by Codex)

# Codex can now access your GitHub repos , PRs , Issues
[mcp_servers.github]
url                  = "https://api.githubcopilot.com/mcp"
bearer_token_env_var = "GITHUB_PAT"


# Codex can now query your PostgreSQL database
[mcp_servers.postgres]
command = "npx"
args    = ["-y", "@executeautomation/database-server", "--postgresql"]








# ════════════════════════════════════
# III)  SETTINGS
# ════════════════════════════════════

# Which AI model Codex uses:
model = "codex-1"

# reasoning level of models 
model_reasoning_effort = "medium"





'''
```
