# 4.. MCP- database-server ( Talk to your DB without writing code) 🗄️


```python
'''


MCP Database Server supports SQLite, SQL Server, PostgreSQL, and MySQL databases..


'''
```

```python
'''

🎯 What is mcp-database-server ?

    👉 DBHub = MCP - Database Server


    Using this MCP Server , you can talk to your database DIRECTLY 
    WITHOUT writing any Python code ! 🤯


    Imagine :
        Instead of writing :
            
            import sqlite3
            conn = sqlite3.connect('users.db')
            cursor = conn.cursor()
            cursor.execute("SELECT COUNT(*) FROM users WHERE signup_date >= '2024-11-01'")
            result = cursor.fetchone()
            ...
        
        
        You just say to Claude in English :
            👉 "Hey Claude , how many users signed up in November ?"
        
        
        & Claude will :
            1. Connect to your DB via DBHub MCP
            2. Translate your English to SQL
            3. Execute the SQL
            4. Give you the answer in plain English 🚀




🎯 Setup Steps :


✅ Step 1 : Open Claude Code terminal in VSCode

        Open VSCode → Open Terminal → type :
        
        👉 claude
        
        ( This starts Claude Code session )



✅ Step 2 : Check existing MCP servers
        
        Inside Claude Code session , type :
        
        👉 /mcp
        
        This shows you ALL the MCP servers currently installed in your system !
        ( Initially this list will be empty )



✅ Step 3 : Go to mcp-database-server GitHub repo

            Search on Google : "mcp-database-server github"
            Installation commands will be given in the README 📖
        



✅ Step 4 : Do the setup

        Install this MCP server using the command provided or u can use 1 single simple command to isntall MCP Server !
        So open New Terminal in VSCode & execute below command !
        
        (  
        ## for SQLite db !
        
        claude mcp add --transport stdio sqlite -- npx -y @executeautomation/database-server "R:\Claude\Files\Sales\databases\amazon_sales_data.db"
        
        )
        
        
        ## PostgreSQL
        claude mcp add --transport stdio postgresql -- npx -y @executeautomation/database-server --postgresql --host localhost --database amazon_sales_data --user postgres --password yourpassword
        
        
        ## MySQL
        claude mcp add --transport stdio mysql -- npx -y @executeautomation/database-server --mysql --host localhost --database amazon_sales_data --port 3306 --user root --password yourpassword
        
        
        ## SQL Server
        claude mcp add --transport stdio sqlserver -- npx -y @executeautomation/database-server --sqlserver --server localhost --database amazon_sales_data --user sa --password yourpassword
        
        
        
        


        ie , now u have configured successfully !
            
            👉 SQLite       🗄️
            👉 MySQL        🐬
            👉 PostgreSQL   🐘
            
            
        Server name: sqlite
        Config saved to: C:\Users\shant\.claude.json
        Project scope: R:\Claude\Files\Sales — meaning this MCP is only active when you run Claude Code from this folder
            
            Means - >> 
            
        Server name: sqlite
        Config saved to: C:\Users\shant\.claude.json
        Project scope: R:\Claude\Files\Sales — meaning this MCP is only active when you run Claude Code from this folder      



✅ Step 5 : Ask simple questions regarding your DB !

        Examples :
            👉 "How many tables are there in my database ?"
            👉 "What is the top selling product?
            👉 "Show all products ranked by total quantity sold

    🚀 Boom — no SQL writing needed !
       Claude handles ALL of it for you !

'''
```

## 5️⃣ AWS MCP Server ☁️


```python
'''

🎯 What is AWS ?

    👉 AWS  =  Amazon Web Services
    👉 AWS  =  World's LARGEST cloud platform 🌍
    

    What is a "cloud platform" ?
    
        Imagine instead of buying & maintaining your own physical servers ,
        you RENT computing power from Amazon !
        
        AWS provides :
            - EC2  →  Virtual machines 💻
            - S3   →  File storage 📦
            - RDS  →  Databases 🗄️
            - Lambda → Serverless functions ⚡
            - CloudWatch → Monitoring 📊
            ... and 200+ other services !



🎯 What does AWS MCP server do ?

    By connecting AWS MCP to Claude Code :
    
        👉 Claude can list your EC2 instances
        👉 Claude can read CloudWatch logs
        👉 Claude can deploy Lambda functions
        👉 Claude can manage S3 buckets
        👉 Claude can check billing & usage 💰
        👉 ... basically operate your entire AWS infrastructure !



🎯 Real-world examples :


    Example 1 : Infra Status
        👉 "Claude , show me all my running EC2 instances and their CPU usage"



    Example 2 : Cost Tracking
        👉 "Claude , give me a breakdown of my AWS bill for this month 
            by service category"



    Example 3 : Debug a Lambda
        👉 "Claude , fetch the last 100 lines of CloudWatch logs for my 
            payment-processor Lambda function and tell me what's failing"


    Example 4 : Deploy
        👉 "Claude , deploy the latest version of my app to AWS Lambda 
            in the us-east-1 region"




🎯 Setup Steps :

    ✅ Step 1 : Go to AWS Console → IAM → Users → Create a new user
    ✅ Step 2 : Create Access Key + Secret Key 🔑
                ( Use FINE-GRAINED permissions !! )
    ✅ Step 3 : Install AWS MCP via Claude Code with the credentials
    ✅ Step 4 : Verify with /mcp
    ✅ Step 5 : Start managing your AWS via Claude ! 🚀




🚨 SUPER CRITICAL Permission Tip 🚨 :

        AWS is connected to YOUR MONEY 💰 !
        A wrong command can :
            - Spin up expensive instances → ₹₹₹ bill 💸
            - Delete production data 😱
            - Expose data publicly via S3 misconfig 🚨
        
        
        🛡️ Best Practices :
            ✅ Create a SEPARATE IAM user for Claude with MINIMUM permissions
            ✅ Start with READ-ONLY access ( use AWS managed policy : ReadOnlyAccess )
            ✅ NEVER use root credentials with Claude 🚫
            ✅ Set up BILLING ALERTS in AWS 🔔
            ✅ Rotate keys every 30 days 🔁

'''
```

## 6️⃣ Docker MCP Server 🐳


```python
'''

🎯 What is Docker ?

    👉 We use Docker to CONTAINERIZE our app …
       so that we can SHARE / DEPLOY our app easily 📦


    What does "containerize" mean ?
    
        Imagine you built a Python app on YOUR Mac …
        Now you want to share it with a teammate on Windows .
        
        Problem :
            - Different Python versions
            - Missing libraries
            - Different OS
            - "It works on my machine !" syndrome 😅
        
        
        Solution = Docker !
            👉 Docker wraps your ENTIRE app + dependencies + environment 
               into a "container" 📦
            👉 This container runs THE SAME WAY on ANY machine !



🎯 What does Docker MCP server do ?

    By connecting Docker MCP to Claude Code :
    
        👉 Claude can list your Docker containers 📦
        👉 Claude can build Docker images
        👉 Claude can write Dockerfiles for your project
        👉 Claude can manage docker-compose setups
        👉 Claude can start , stop , restart containers
        👉 Claude can check container logs 📜



🎯 Real-world examples :


    Example 1 : Generate Dockerfile
        👉 "Claude , my project uses Python 3.10 + FastAPI + Redis . 
            Generate a production-ready Dockerfile + docker-compose.yml"



    Example 2 : Debug a container
        👉 "Claude , my api-server container keeps restarting . 
            Read the logs and tell me what's wrong"



    Example 3 : Cleanup
        👉 "Claude , list all stopped containers and unused images , 
            and clean them up to free space"




🎯 Setup Steps :

    ✅ Step 1 : Make sure Docker is installed on your system 🐳
    ✅ Step 2 : Install Docker MCP server via Claude Code
                ( github.com/ckreiling/mcp-server-docker or similar )
    ✅ Step 3 : Verify with /mcp
    ✅ Step 4 : Start asking Claude about Docker ! 🚀




🚨 Permission tip :
        ✅ Read commands ( list containers , read logs ) → SAFE
        🚨 Be careful with auto-approve on : docker rm , docker kill
        🔥 NEVER auto-approve : docker system prune -a ( deletes everything ! )

'''
```
