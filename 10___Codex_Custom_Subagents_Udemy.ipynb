```python
'''

Hi there 👋 , Welcome to one of the MOST powerful sections of this entire course :

👉 CUSTOM SUB-AGENTS 🔥


------------------------------------------------------------
🔹 Quick Recap  -  What we have learnt so far
------------------------------------------------------------


      [diagram]

                    We have already learnt :
                              |
                  ------------+--------------------
                  |           |                   |
            Why to use     What are        Built-in sub-agents
            sub-agents ?   sub-agents ?    ( explorer , worker ,
                                            default )
                              


------------------------------------------------------------
🔹 This Module Goal
------------------------------------------------------------

👉 In this Module ,  we will understand :
      what are CUSTOM sub-agents , 
      need for Custom Sub-agents 
      Skills vs Custom Sub-agents
      [Demo] : Create sub-agent for our workflow 
      How to CALL / TRIGGER them ..
      7 keys of Custom Sub-Agent





They sit RIGHT BESIDE the built-in ones :

      [diagram]
                            🤖  Sub-agents
                                  |
                  ----------------+----------------
                  |                                |
          Built-in Sub-agents              Custom Sub-agents
          (given by Codex)                 (built by YOU)
          - explorer                        - sales-data-analyst
          - worker                          - security-auditor
          - default                         - blog-post-writer
                                            - jira-ticket-writer
                                            - ... anything YOU want






------------------------------------------------------------
🔹 So , what exactly are Custom Sub-agents ?
------------------------------------------------------------

👉 Custom Sub-agents are the agents that WE make ourselves ,
   for OUR own needs , for OUR own workflows ..

In simple words :

      Custom Sub-agent  =  Your OWN AI specialist
                           +  designed by YOU
                           +  for a SPECIFIC repeating task in your project




------------------------------------------------------------
🔹 Real-world Examples
------------------------------------------------------------




------------------------------------------------------------
🔹  Example  →  amazon_sales_customer.sqlite
------------------------------------------------------------

1)  Imagine you want to do a DATA QUALITY AUDIT of your database ..
   👉 Just to check : " Is our customer & sales data clean ,
                        safe , and report-ready or not ? "

So we basically have to check :

      [diagram]

      amazon_sales_customer.sqlite
         │
         ├── customer data  ──►  Is PII ( emails , phone numbers )
         │                       properly masked before reports go out ?
         │
         ├── sales records  ──►  Are there NULL values , negative prices ,
         │                       or duplicate orders hiding in the data ?
         │
         └── product data   ──►  Are GST rates correctly mapped
                                  to each product category ?


This is a serious , detail-heavy task ..

👉 To do this audit , Codex basically has to LOAD your entire database
   schema & records into the context window & analyse it ..

👉 And THAT is exactly where you can use a Custom Sub-agent 🔥


------------------------------------------------------------
🔹 Wait .. cant the Built-in agent do this ?
------------------------------------------------------------

   👉 The above task CAN be done easily using a Built-in agent .. ✅
   👉 BUT  ->>  when you are working at a real company ,
       that company has its OWN guidelines / checklist
       for data quality audits ..

   Eg : Amazon might say :
       - " Customer emails must NEVER appear in reports —
           always mask as  ****@****.com "
       - " Revenue must ONLY be calculated as  Quantity × Price —
           no rounding , no approximations "
       - " Any product NOT in our GST rate table must be
           FLAGGED — never assume a default rate "
       - " Every audit report must include the table name ,
           row count , and date range it was run on "

   The Built-in agent does NOT know any of this !
   It only knows GENERIC data quality best practices ..

👉 Thats where you HAVE to create a CUSTOM Sub-agent ..
   which will follow YOUR company's specific data guidelines ✅


------------------------------------------------------------
🔹 Summary  ->>   Built-in  vs  Custom
------------------------------------------------------------

   👉  Built-in agent
         - has GENERIC knowledge of doing things
           ( eg : checks for NULLs , duplicates , basic data issues )
         - does NOT know about YOUR company data rules ❌


   👉  Custom Sub-agent
         - has SPECIALISED knowledge of doing things
         - WILL know about your company rules  ✅

               eg :  mask PII before reporting
                     flag unknown GST products
                     exact column naming contract
                     report format & lineage rules












U can create various custom-agent as per your workflow 


   👉  sales-data-analyst   ->>  cleans + analyses sales CSVs in YOUR format
   👉  jira-ticket-writer   ->>  writes tickets in YOUR company's template
   👉  security-auditor     ->>  scans code following YOUR company's checklist
   👉  resume-reviewer      ->>  ATS-style feedback in YOUR scoring rules
   👉  blog-post-writer     ->>  writes blogs in YOUR tone / voice



& every custom agent will have its own :
    diff model access
    own set of TOOLS / MCP servers
    own SYSTEM PROMPT
    own set of SKILLS


& Each one is a SPECIALIST .. designed for one specific job .. ✅






But now you may ask -
" Sir , we ALREADY have skills designed to carry out a specialised task , then 
        WHY do we even need custom sub-agents ? "

    This is something which we will discuss in next session !

'''
```

# 2.. Skills vs Custom-Agents ! ( when to use What)

### for above use-case !

```python
'''

Hi there 👋 , Now that we know WHAT Skills & Custom Sub-agents are ..
    & both are designed to perform specialised tasks !
    Now the que is : When to use what ?
                     What are differences between both ?

        
        
        
------------------------------------------------------------
🔹 Quick Recap  ( 1 line each )
------------------------------------------------------------

      Skill             =  a MANUAL 📋pure markdown file ..
                           gives the codex ie  HOW-TO perform certain tasks ..
                           but instructions/rules are soft — ie codex will read it,follow it but sometimes it may overwrite it

      Custom Sub-agent  =  a WORKER   🧑‍💼
                           instructions are written inside  .toml file ..
                           but instructions/rules are hard — ie codex will always follow these rules !





------------------------------------------------------------
🔹 The Decision Rule  ( dead simple )
------------------------------------------------------------

      [diagram]

            Ask yourself ONE question :

                  " Do I need SYSTEM-LEVEL CONTROL
                    over HOW this task runs ? "
                          |
              ------------+------------
              |                       |
             NO                      YES
              |                       |
           Use a                  Use a
           SKILL  📋          Custom Sub-agent  🧑‍💼


      System-level control  means :
            →  lock to read-only sandbox
            →  specific model / reasoning effort
            →  isolated context window
            →  restricted MCP tools













------------------------------------------------------------
🔹 Real Example  →  Data Quality Audit
        ( amazon_sales_customer.sqlite )
------------------------------------------------------------

Imagine you want to do a  DATA QUALITY AUDIT  of your database ..
👉 Just to check :  " Is our customer & sales data clean ,
                      safe , and report-ready or not ? "

      [diagram]

      amazon_sales_customer.sqlite
         │
         ├── customer data  ──►  Is PII ( emails , phone numbers )
         │                       properly masked before reports go out ?
         │
         ├── sales records  ──►  Are there NULL values , negative prices ,
         │                       or duplicate orders hiding in the data ?
         │
         └── product data   ──►  Are GST rates correctly mapped
                                  to each product category ?


👉 This is a  SERIOUS  task ..
   Codex has to load the ENTIRE database schema into context & analyse it ..


------------------------------------------------------------
Now u may ask  →  " Can a SKILL handle this ? "
------------------------------------------------------------

Lets go row by row from our table above  👇


---
Row 1  →  Lock to read-only sandbox ( it means Codex can only read this database , it can't edit database)
---

👉 In SKILL.md you can write :

      ## Rules
      - Do NOT modify  amazon_sales_customer.sqlite
      - Read only — never write

   BUT — thats just a TEXT INSTRUCTION ( a polite request ) ..
   Codex COULD still write to the DB if something overrides it ..
   One wrong move  →  your entire customer database is corrupted  😱  ❌


👉 With a Custom Sub-agent :

      sandbox_mode = "read-only" in data-quality-auditor.toml

   Codex cannot write to the DB ..
   It is — a SYSTEM-LEVEL LOCK  ✅
   Here you are providing commands basically !


---
Row 2  →  Set a specific model + reasoning effort
---

👉 In SKILL.md you can write :

      ## Instructions
      - Use high reasoning effort
      - Use the most capable model

   BUT — thats just a TEXT INSTRUCTION ..
   The agent still runs on whatever model your session uses ..
   A cheap fast model doing a deep DB audit  →  misses critical issues  ❌


👉 With a Custom Sub-agent :

      model                  = "gpt-5.4"
      model_reasoning_effort = "high"

   Codex ALWAYS uses this exact model + depth for the audit ..
   No maybes , no inheriting the wrong session model  ✅


---
Row 3  →  Run in its own isolated context window
---

👉 With a SKILL — the audit runs INSIDE your main session ..
   Thousands of tokens of DB schema + audit findings
   pile up in YOUR main context ..
   Now your main session is slow , confused , token-heavy  ❌

👉 With a Custom Sub-agent — the audit runs in its OWN private context ..
   Findings stay isolated ..
   Main session stays clean  ✅


---
Row 4  →  Specific MCP tools only
---

👉 In SKILL.md you can write :

      ## Tools
      - Use ONLY the SQLite MCP tool
      - Do NOT use GitHub or any other tool

   BUT — thats just a TEXT INSTRUCTION ..
   The agent still has access to ALL MCPs in config.toml ..
   It could accidentally pull data from GitHub or other sources  ❌

👉 With a Custom Sub-agent :

      [mcp_servers.sqlite]
      command = "sqlite-mcp-server"

   ONLY the SQLite tool exists for this agent ..
   It CANNOT access any other MCP — even if it tries  ✅







------------------------------------------------------------
🔹 So the Custom Agent for this task looks like :
------------------------------------------------------------

      # .codex/agents/data-quality-auditor.toml
      # now below rules will be locked for data-quality-auditor "AGENT" !

      name                   = "data-quality-auditor"
      model                  = "gpt-5.5"              ←  Row 2 ✅
      model_reasoning_effort = "extra-high"                 ←  Row 2 ✅
      sandbox_mode           = "read-only"            ←  Row 1 ✅

        [mcp_servers.sqlite] 
        command = "npx" 
        args = ["-y", "@executeautomation/database-server", 
                "R:/Codex/Codex_Masterclass/MCP/Sales_Analysis/data/amazon_sales_customer.sqlite"]

      """


      Each TOML key  →  enforces ONE row of our table  🔥


------------------------------------------------------------
🔹 Skill  vs  Custom Sub-agent  →  for THIS task
------------------------------------------------------------

| Feature                           | Skill           | Custom Sub-agent     |
|-----------------------------------|-----------------|----------------------|
| Knows audit rules & GST rates     | ✅ ( but soft ) | ✅ ( hard )          |
| DB can NEVER be modified          | ❌ suggestion   | ✅ sandbox_mode      |
| Uses gpt-5.4 + high reasoning     | ❌ suggestion   | ✅ model key         |
| Audit runs in isolated context    | ❌              | ✅ own context       |
| Only SQLite MCP — nothing else    | ❌ suggestion   | ✅ mcp_servers       |


------------------------------------------------------------
🔹 The ONE Rule to remember
------------------------------------------------------------

      In case of Skill   →  ( rules are SOFT  — agent may ignore )
      In case of Agent   →  ( rules are HARD  — system always enforces )

      For a  DATA QUALITY AUDIT  on a production database —
      you CANNOT afford soft instructions  🔥
      You need a Custom Sub-agent  ✅
      
'''
```

# 3.. Where are Custom Sub-agents Stored ?  📂


```python
'''

Hi there 👋 , Very practical question now :


    [ SHOW in VS CODE ]
   "OK , I want to BUILD a custom sub-agent ..
    WHERE does Codex actually expect this file ?"


Lets answer it step by step 🚀


------------------------------------------------------------
🔹 The Storage Convention
------------------------------------------------------------

👉 Every custom sub-agent lives as a single .toml file ( one agent per file ) inside :
      .codex / agents /  <agent-name> .toml

   ( convention : match the filename to the agent's  name  field )


   So the structure looks like :

      [diagram]

           .codex /                       ← Codex's config folder
              └── agents /                ← sub-folder for ALL custom agents
                     ├── data-quality-auditor.toml
                     ├── code-reviewer.toml
                     └── blog-writer.toml




------------------------------------------------------------
🔹 What does ONE custom agent file actually look like ?
------------------------------------------------------------

👉 A Codex custom agent is a single TOML file where the CONFIG keys &
   the INSTRUCTIONS live together  ( there is NO separate markdown body ) :


      [diagram]

           agent .toml file
                  |
         ---------+----------
         |                  |
         |                  |
   config keys        developer_instructions
   ( name , model ,    ( the system prompt =
     sandbox_mode .. )    HOW it should work )


👉 config keys             ->>   the CONFIG of your agent
                                 ( name , description , model , sandbox_mode , etc. )

                                name = using this name , u can this Agent !
                                description = Human-facing guidance for when Codex should use this agent
                                model =  Specific model this agent uses (e.g. gpt-5.4, gpt-5.4-mini, gpt-5.3-codex-spark)
                                sandbox_mode = read-only or workspace-write
                                    ie whether Agent can only read workspace or edit workspace or complete access to system



👉 config keys              ->>  WHAT the agent is + WHAT it can touch
👉 developer_instructions   ->>  HOW it performs the task






👉 developer_instructions  ->>   detailed instructions for the sub-agent
                                 ie ..  HOW it should perform its task ✨


Example layout :

      # ~/.codex/agents/data-quality-auditor.toml

      
      name                   = "data-quality-auditor"
      description = "Use this agent for company-grade data-quality audits."
                    Triggers on : "audit" , "data quality" , "check database" ,
                                   "PII check" , "GST validation" .
      model                  = "gpt-5.4"              ←  Row 2 ✅
      model_reasoning_effort = "high"                 ←  Row 2 ✅
      sandbox_mode           = "read-only"            ←  Row 1 ✅



      developer_instructions = """
      You are a data quality auditor for Amazon India .  ←  Row 3 ✅
      You run in your OWN context — never pollute main session .

      STRICT rules :
        - NEVER expose raw emails or phone numbers
          mask as  ****@****.com
        - Revenue  =  Quantity × Price  ONLY
        - Flag ANY product NOT in GST rate table — never guess
        - Use EXACT column names : Prod_name , Quantity , Price , region
      """
     








------------------------------------------------------------
🔹 The 2 TYPES of Custom Sub-agents ( based on location )
------------------------------------------------------------

      [diagram]

                          Custom Sub-agents
                                  |
                   ---------------+---------------
                   |                              |
            User / personal-              Project-level
            level                          ( also called : repo-level )

            Inside  HOME directory         Inside  PROJECT directory
            of .codex                      of .codex

            ie .. ~/.codex/agents/         ie .. ./.codex/agents/
                       *.toml                        *.toml



👉 USER-LEVEL  ( system-wide )

   Path  ->>  ~/.codex/agents/<name>.toml
   Available to ALL projects on your machine ✅
   
       There is no agents/ folder provided by Codex..
       You have to create it yourself — manually.




👉 PROJECT-LEVEL  ( scoped to ONE project )

   Path  ->>  <your-project>/.codex/agents/<name>.toml

   ie ..
        /my-sales-project/
            ├── .codex/
            │      └── agents/
            │             ├── sales-data-cleaner.toml
            │             └── eda-runner.toml
            ├── src/
            └── ...

   Available ONLY inside THIS project ✅
   ( project-scoped agents inherit the parent session's settings unless
     they override them )








In the next section , we will deep-dive into the Hands-on to create sub-agent !

'''
```

# 5.. [Live Demo] : The 7 Keys of a Custom Sub-agent  🗝️
        First Show VS CODE , then PPT , then back to VS CODE


```python
'''

------------------------------------------------------------
🔹 Full Example  →  All Keys Together
------------------------------------------------------------

A complete custom sub-agent.toml file for our
sales_data_analysis project :

       # .codex/agents/data-quality-auditor.toml

       name        = "data-quality-auditor"
       description = """
       Use this agent to audit data quality of amazon_sales_customer.sqlite .
       Triggers on : "audit" , "data quality" , "check database" ,
       "validate data" , "customer data check" .
       """
       model                  = "gpt-5.5"
       model_reasoning_effort = "extra-high"
       sandbox_mode           = "read-only"
       nickname_candidates    = ["DataAuditor", "QualityBot"]

        [mcp_servers.sqlite]
        command = "npx"
        args = ["-y", "@executeautomation/database-server", "R:/Codex/Codex_Masterclass/MCP/Sales_Analysis/data/amazon_sales_customer.sqlite"]


       [[skills.config]]
       path    = "path_to_skill"
       enabled = true

       developer_instructions = """
       You are a senior data quality auditor for Amazon India .
       Your ONLY job is to audit  data/amazon_sales_customer.sqlite
       and report findings . NEVER modify any records .

       DATABASE STRUCTURE :
         Table   :  amazon_sales
         Rows    :  500
         Columns :
           user_name      ←  TEXT    ( customer identifier — all lowercase )
           Prod_name      ←  TEXT    ( product name — capital P )
           Quantity       ←  INTEGER ( units sold — capital Q )
           Price          ←  INTEGER ( unit price — capital P )
           Sales_Channel  ←  TEXT    ( Website / Wholesale / Amazon — capital S , C )
           Customer_Type  ←  TEXT    ( New / Returning — capital C , T )

       STRICT AUDIT RULES :
         1. Column names are CASE-SENSITIVE — use EXACT names above
            ( never write  prod_name  or  sales_channel  or  customer_type )

         2. Revenue  =  Quantity × Price  ONLY
            no rounding , no approximations

         3. user_name is a customer identifier
            NEVER expose or group by raw user_name in summary reports

         4. Flag ANY product NOT in our GST rate table — NEVER guess a rate :
              Electronics  →  18%
              Clothing     →   5%
              Food         →   0%
              Books        →   5%

         5. Check and report on :
              ✅  NULL values in any column
              ✅  Negative or zero  Quantity / Price
              ✅  Duplicate orders ( same user_name + Prod_name + Price )
              ✅  Sales_Channel values outside : Website , Wholesale , Amazon
              ✅  Customer_Type values outside : New , Returning

         6. End every report with :
              - Table name
              - Total row count
              - Count of flagged rows
              - Severity-ranked summary table
       """




------------------------------------------------------------
🔹 Each key  →  doing its job
------------------------------------------------------------

      name              →  "data-quality-auditor"
                            unique identity — you call it by this name  ✅

      description       →  tells Codex WHEN to use this agent
                            ( audit , PII check , GST validation .. )  ✅

      model             →  gpt-5.5
                            most capable — deep DB analysis needs it  ✅

      model_reasoning   →  extra-high
      _effort               serious audit — needs deep thinking  ✅

      sandbox_mode      →  read-only
                            PHYSICALLY cannot touch the database  ✅

      mcp_servers       →  sqlite ONLY
                            cannot accidentally call GitHub or other tools  ✅

      skills.config     →  data-quality-checklist SKILL attached
                            agent already knows your company audit rules  ✅

      nickname          →  "DataAuditor" / "QualityBot"
      _candidates           you instantly know which agent is running  ✅

      developer         →  PII masking , GST flagging , exact column names ,
      _instructions         report format — all HARDCODED  ✅


------------------------------------------------------------
🔹 Quick Summary of Keys
------------------------------------------------------------

      🗝️  Key                   |  What it controls                          |  Required ?
      --------------------------|--------------------------------------------|--------------
      name                      |  unique identity of the agent               |  ✅ Yes
      description               |  guidance for WHEN to use this agent        |  ✅ Yes
      developer_instructions    |  the system prompt ( how to do the task )   |  ✅ Yes
      model                     |  gpt-5.5 / gpt-5.4-mini / spark model       |  optional
      model_reasoning_effort    |  low / medium / high / extra-high           |  optional
      sandbox_mode              |  read-only / workspace-write                |  optional
      mcp_servers               |  which MCP tools the agent gets             |  optional
      skills.config             |  attached how-to skills                     |  optional
      nickname_candidates       |  display names for the spawned agent        |  optional


Master these keys  ->>  and you can build ANY kind of specialist
sub-agent for ANY project 🔥


And in the next section , we will see  ->>  HOW to actually
CALL these custom sub-agents inside Codex 📞

'''




'''
```
