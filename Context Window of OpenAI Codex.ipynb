# 2.. How the Context Window of Codex Works ⚙️


```python
'''


Hi there 👋 , in the last session ..
We understood WHAT a Context Window is .

Now let's understand : 👉 HOW does it actually work inside Codex ?



━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
👨‍🏫  INSTRUCTOR EXPLANATION
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

📌 IMPORTANT POINTERS about Codex models :

    👉 In Codex today , a GPT-5-Codex class model has a context window of around 400K tokens
                         and some models can run with context window as large as 1M tokens ..

    👉 Every time , u start a new conversation with Codex , 
        fresh context window will be around 400K tokens
        

    👉 Models have a limit of 400K tokens & they will be consumed by your questions + CODEX REplies
        Your query ->> I/P tokens
        Codex replies ->> Output Tokens


💡 One more practical note :
    Codex replies ( output tokens ) often cost MORE than your questions ( input tokens ) ,
    because the reply includes generated code , explanations , and the results of tools it runs .






Now here comes the MOST IMPORTANT concept of this whole section 🔥 :

    👉 LLMs are STATELESS — they do NOT have memory of their own !

        What does "stateless" mean ?
            👉 It means the model does NOT actually remember your previous messages on its own .

        So then .. how does Codex "remember" your conversation ? 🤔

        👉 Here is the trick :
           Every time you send a new message , Codex SECRETLY re-sends the ENTIRE conversation so far
           ( all your previous questions + all the model's previous replies ) back to the model
           as INPUT tokens .

        👉 That is the ONLY way the model can "remember" — by re-reading everything each turn .

⭐ THIS is the reason your context window fills up SO fast !


Let's have a rough estimate of count tokens turn-by-turn .
Imagine each of your messages and each Codex reply costs about 100 tokens :

    Turn 1 :   [ you ] [ codex ]                                  →  ~200 tokens
    Turn 2 :   [ you ] [ codex ] [ you ] [ codex ]                →  ~400 tokens
    Turn 3 :   [ you ] [ codex ] [ you ] [ codex ] [ you ][codex] →  ~600 tokens
       ..
    Turn 10 :  ( all previous conversation re-sent ) ..           →  ~2000 tokens

👉 See the pattern ?
   Every new turn drags ALL the previous turns along with it .

    🔑 Conclusion :
        Bigger the size of your session / conversation ,
        HIGHER the chance your context window gets EXHAUSTED .









━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
🌍  REAL-WORLD EXAMPLE
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Let's say you start a Codex session to collect amazon users from 5 Sources  !
[Show VSCODE]


    1 source ->> CSV
    2 source ->> Database
    3 Source ->> Big Data Sources
    4 source ->> WEB Scraping
    5 source ->> JSON
    
    So u can create multiple sessions ( 1 session for each source) else your limit will be exhausted !
    



━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
🎯  STUDENT TAKEAWAY
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

    👉 Context window is a TOKEN limit ( ~400K for current GPT-5-Codex models ; effective is a bit lower ) .
    👉 LLMs are STATELESS — Codex re-sends the WHOLE conversation each turn so the model can "remember" .
    👉 That is exactly WHY long sessions fill the window fast .
    👉 Good practice ( we'll prove this later ) :  ONE focused session / thread per feature .


'''
```

# 3.. Components of the Codex Context Window ( in detail ) 🧩


```python
'''


Hi there 👋 , in the last session ..
We understood HOW the context window works , and why it fills fast .

Now let's open it up and see : 👉 WHAT exactly is INSIDE the Codex context window ?


━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
🎙️  NARRATION SCRIPT
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Here is a very important idea first :

    👉 A model may have a 400K window ,
       but the window you can ACTUALLY use is a little smaller .

It's just like buying a laptop with 8GB of RAM 💻 ..
        👉 the OS already eats some of it , so the USABLE RAM is less than 6 or 7GB .

Same with Codex : some of the window is ALREADY occupied the moment your session starts ,
and some is RESERVED in advance . 


━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
👨‍🏫  INSTRUCTOR EXPLANATION  —  the layout of the window
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

🧩 Codex Context Window ( who lives where ) :

    ┌───────────┬──────────┬──────────┬───────────────────┬──────────┬───────────┬───────┐
    │  System   │  Tool    │ AGENTS   │   Conversation    │  Tool    │  Auto-    │ Free  │
    │  Prompt   │  Schemas │  .md     │     History       │ Results  │  Compact  │       │
    └───────────┴──────────┴──────────┴───────────────────┴──────────┴───────────┴───────┘
       (system instructions)   (project memory)              (Skills + MCP)   (reserved)


Let's go through each tenant one by one 👇


1) 🧠  System Prompt
        👉 When you start a session , Codex loads a set of built-in instructions ( the system prompt ) .
        👉 These instructions are designed by OpenAI & This takes up some tokens before you even type anything .

2) 🛠️  Tool Schemas
        👉 Remember :  Codex = LLM + TOOLS .
        👉 Every tool ( read file , edit file , run command , search , etc. ) has a SCHEMA :
               - the definition of the tool
               - what the tool can do
               - what INPUTS the tool needs
               - what OUTPUTS the tool returns
        👉 All these schemas sit inside the window and cost tokens .

3) 📄  AGENTS.md   ←  ( this is Codex's project-memory file ! )
        👉 AGENTS.md is loaded at the START of your session .
        👉 It is like an "overview" file for your project .
        👉 Inside it you keep : coding standards , conventions , what to do , what NOT to do .


4) 💬  Conversation History
        👉 As we learned , this is your messages + Codex's replies , re-sent every turn .
        👉 This is usually the FASTEST-growing part
        
        
5) 📤  Tool Results
        👉 Every tool Codex runs produces some output ( file contents , command logs , test output ) .
        👉 This output is often VERBOSE , and it ALSO needs tokens to live in the window .

6) ⚡  Skills + MCP tools
        👉 Skills ( your reusable task playbooks ) and MCP servers ( external connectors )
           also consume tokens — their descriptions and results live in the window too .

7) 🗜️  Auto-Compaction Reserve
        👉 Codex keeps a chunk of the window RESERVED for compaction ( we'll cover this next session ) .

8) 🟢  Free
        👉 Whatever is left over — this is the room you ACTUALLY have to work in .





In the next session , we'll understand : 👉 What is Auto-Compaction , and how Codex uses it
to keep your long sessions alive 🗜️

'''
```

# 4.. Auto-Compaction in Codex 🗜️


```python
'''


Hi there 👋 , in the last session ..
We saw every component inside the context window !

Now let's understand : 👉 What is Auto-Compaction , and why does Codex do it ?


━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
🎙️  NARRATION SCRIPT
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

    Around 60-70k Tokens are reserved for Auto compaction !
    Its' like 

    Context window usage :
    [████████████████████░░░░]   ← imagine you're at 80–90% of Context Window usage , say ~350K tokens used


So what does Codex do to save you ?
        👉 It COMPACTS the context — it summarises the conversation so far into a much
           shorter form & store it into a Auto-compaction buffer !

When Codex does this AUTOMATICALLY , we call it AUTO-COMPACTION .




[VS CODE]  show : 
    Imagine I am using Codex 
    my window is ~80–90% full ? 
    run : /status in VS CODE CODEX Chat panel

    How do I handle it ?"

    There are 2 ways to do AUTO-COMPACTION : 


    a)  Let Codex handle it AUTOMATICALLY 
            👉 Codex summarises conversation & store it into Auto Compaction Buffer & makes your token usage free !
            👉 But remember : it's not loss-less( it means u will loose some essence of info while summarising) 
                    , and Codex may trigger LATE .
            
        But when Auto-compaction buffer fills , Codex will not allow u to chat, degrades response quality 
        So close that session & create a new one !
        
        

    b)  Handle it MANUALLY with the  /compact  command  ⭐ ( the better habit )
            👉 You decide WHEN to compact , before quality of Codex drops .
            👉 run /compact command !







'''
```

#### Why the Context Window Matters 💸


```python
'''


Hi there 👋 , in the last session ..
We learned how Codex rescues long sessions with auto-compaction .

Now let's understand : 👉 WHY should you even care about the context window ?
There are TWO big reasons : COST and QUALITY .


━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
🎙️  NARRATION SCRIPT
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Reason 1 — 💰  COST

    👉 Remember : every turn re-sends the whole conversation as input tokens .
    👉 More tokens in the window  →  more tokens processed every turn  →  HIGHER cost .

    So a bloated context window is literally burning your usage / your money .


Reason 2 — 📉  QUALITY DEGRADATION

    👉 As your context window FILLS UP , the OUTPUT QUALITY of Codex starts to DROP .

━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
👨‍🏫  INSTRUCTOR EXPLANATION
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Let me show you the danger zone visually :

    Context window usage :
    [████████████████████░░░░]   ← imagine you're at 80–90% , say ~120K tokens used

    Now , at this point , if you give Codex a brand-NEW task ..
        👉 the output QUALITY will likely DECREASE .
        👉 responses also get SLOWER , because the model is chewing through more tokens per turn .
        👉 So compress the conversation !




'''
```

# 5.. Good Practices to Handle the Context Window 🏆


```python
'''


━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
👨‍🏫  INSTRUCTOR EXPLANATION  —  THE GOOD-PRACTICES CHECKLIST
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Imagine you're an AI Engineer / Data Analyst working on an Amazon Product Analytics platform.

Your task is to collect and analyze data from multiple sources:

📄 CSV files
🗄️ SQL Databases
☁️ Big Data Sources (Spark / Hadoop)
📦 JSON APIs
🌐 Web Scraping

And you need columns like:

    Product Name
    Price
    Discount %
    Ratings
    Reviews
    User Information
    Product Category
    Seller Information
    Inventory Status




✅ 1) ONE thread / session per feature

    ✅ GOOD: Create separate sessions: 
    Session 1 → CSV Data Ingestion 
    Session 2 → Database Pipeline 
    Session 3 → Web Scraping Pipeline 
    Session 4 → Review Sentiment Analysis 
    Session 5 → Dashboard Development 
    
    👉 Every new session starts with a fresh context window. 
    👉 Faster responses. 
    👉 Lower token usage. 
    👉 Better code quality.
    
    
    
    
✅ 2) Check /status periodically and /compact proactively

    Your window is filling quickly. 
    Run: /status to check context usage. 
    
    Before reaching the danger zone: /compact
    
    
    
✅ 3) Write focused, polished, SPECIFIC prompts

    ❌ BAD:

            Analyze my Amazon data.
            
            
            
    ✅ GOOD: Extract average product ratings, total reviews, and price distribution from amazon_reviews.csv.
    
    👉 Precise prompts = fewer tool calls. 
    👉 Fewer tool calls = lower token cost. 
    👉 Lower token cost = longer sessions.






✅ 4) Use SUBAGENTS for parallel / heavy work 🤖

    Imagine your Amazon project contains:

            20 million reviews
            5 data sources
            multiple ETL pipelines

    Instead of doing everything in one window:

    Create specialized subagents.

    🤖 Data Extraction Agent

            Collects:
                - CSV data
                - Database records
                - JSON feeds

    🤖 Web Scraping Agent

            Extracts:
                - product prices
                - ratings
                - reviews

    🤖 Sentiment Analysis Agent

            Processes:
                - customer reviews
                - review sentiment
                - complaint detection



Every subagent gets its OWN context window. 

So: Main Agent Window 
            ↓ 
    delegates work 
            ↓ 
    Subagent Window #1 
    Subagent Window #2 
    Subagent Window #3




'''
```
