# 1.. How LLM actually Works ( Behind the scenes ) ?


```python
'''

Hi there 👋 , Welcome to one of the MOST IMPORTANT sections of this Claude Code course !

In this section , we will deep dive into ->> 
    Subagents
    Their types 
    How they work behind the scenes 
    How they get triggered & also a full hands-on demo !




So lets first understand something very basic :

------------------------------------------------------------
🔹 First , lets understand : How does an LLM actually work ?
------------------------------------------------------------

we don’t talk to LLM directly ..

We talk to LLMs via ->> 👉 API Call

[diagram]

There is a wrapper (API) around the LLM .

Step by step :
   👉 We send our query to API
   👉 API feeds that query into LLM
   👉 LLM gives a response 
   👉 Then we receive that response back via API

Thats how every LLM in this world works …. simple ✅


------------------------------------------------------------
🔹 But here is the BIG problem with LLMs
------------------------------------------------------------

👉 LLMs are STATELESS / MEMORY-LESS .
   They DO NOT have memory of past conversations on their own !


Lets see an example :

API Call 1 :
   You ask  ->> "What is capital of India ?"
   LLM     ->> "Delhi"   ✅


API Call 2 :
   You ask  ->> "What about Germany ?"
   LLM     ->> "What about Germany in what context ?" 🤯

   Why ?
   Because LLMs don’t have memory ..
   They don’t know what you asked in Call 1 !


------------------------------------------------------------
🔹 So how did Claude developers fix this ?
------------------------------------------------------------

They found a solution :
👉 When you ask a question to LLM ,
   we don't just send the new question -
   we send  ->>   new query + entire previous conversation

So now LLM will have full context of what was discussed before ✅

This is the trick that gives ChatGPT , Claude , Gemini their "memory feel" !


            new query 
                +                   ----->  LLM   (now has full context)
        entire previous conversation


'''
```

```python
'''

Let Understand it using a Real World Example : 

------------------------------------------------------------
🔹 Real Example  ->>  Imagine you are a Data Analyst @ YouTube
------------------------------------------------------------

You are using Claude Code to do your day to day work ..


TURN 1 :
   You say  ->> "With the help of LLM, collect YouTube data for me .."
   Lets say  ->>  this turn consumed roughly ≈ 30K tokens

   ie till Turn 1  ≈  30K tokens spent ✅


TURN 2 :
   Now you say  ->> "Clean this data .."

   But remember -  Claude has to send :
        👉  query  ( Clean this data )
                +
        👉  previous Turn 1 conversation  ( 30K tokens )
              
   Total sent to LLM  ≈  30K tokens
   LLM does the cleaning & returns new cleaned output  ( imagine 5K tokens )

   ie till Turn 2  ≈  60K – 65K tokens have been spent .. 😳


TURN 3 :
   Now you ask LLM  ->> "Perform Data Analysis on it"

   Again Claude sends  ->>  query + previous conversation [ 65K ]
   LLM responds with EDA output  ( imagine ≈ 20K tokens )

   ie till Turn 3  ≈  130K tokens  +  20K (EDA O/P)
                  =  150K tokens spent !! 🤯


You can clearly see ..
👉 Your context window is getting FILLED so fast ..
👉 And each call is costing you MORE money than the last one ..



------------------------------------------------------------
🔹 So there are 2 MAJOR problems with this approach
------------------------------------------------------------

PROBLEM 1 :  Context Window Overflow
   👉 Every model has a fixed context window ( like 200K , 1M etc )
   👉 If you keep stacking conversation , it fills very fast ..
   👉 Once full -> LLM forgets or breaks ..


PROBLEM 2 :  Lost in the Middle Effect
   👉 LLMs are known to pay MOST attention to the
      START  &  END  of the context .
   👉 They tend to MISS the MIDDLE tokens ..
   👉 So even if your context fits , LLM might "skip" important details
      sitting in the middle ..




------------------------------------------------------------
🔹 So whats the solution ?
------------------------------------------------------------

👉 This is EXACTLY why the idea of  ->>  SUBAGENTS  came into picture 🚀

Subagents solve BOTH of these problems in one shot ..
And thats what we are going to learn next !

'''
```

# 2.. What is a Subagent ?


```python
'''

Now we know the problem ..
👉 Context window fills too fast
👉 Lost in the middle effect

Lets see how Subagents solve this beautifully ✨


------------------------------------------------------------
🔹 Definition
------------------------------------------------------------

👉 Subagents are AI Assistants that run inside their OWN
   ISOLATED Context Window ..



   Sub-agent  =  Mini AI agent  +  its own SEPARATE context window


------------------------------------------------------------
🔹 How does it actually work ?
------------------------------------------------------------

[diagram]


                          🟦  Main Agent  🟦
                                 |
              -------------------+-------------------
              |                  |                  |
        Sub-agent 1         Sub-agent 2        Sub-agent 3
        (Data Collector)    (Cleaner)          (Analyser)
        (own context win)   (own context win)  (own context win)



Lets understand step by step :

👉 You give a task to the Main Agent ( the main Claude session )
👉 Main agent says  ->> "Hmm , this part of the task is heavy ,
   let me delegate it to a sub-agent .."
👉 It spins up a Sub-agent  ( for eg : a Data Collection agent )
👉 That Sub-agent has its OWN isolated context window
👉 It does its work independently
👉 Finally , it RETURNS only the result back to the Main agent

So now Main agent gets :
   👉 only the FINAL collected data



------------------------------------------------------------
🔹 And the magic part is …
------------------------------------------------------------

👉 Once the sub-agent's task is done ->>
   its ENTIRE context window is DELETED ✅

Main agent stays clean .
Main context window stays light .
No bloat , no overflow , no lost-in-the-middle ✨


------------------------------------------------------------
🔹 What is the advantage of this ?
------------------------------------------------------------

Q : What is adv. of this ?

Ans : 
   👉 Your context window will NOT fill so fast
   👉 Which automatically results in  ->> 
     👉  LOWER COST 💰 + 👉 better quality responses + 👉 cleaner separation of work


Thats actually the MAGIC of subagents .



'''
```

# 3.. Why use Subagents ? ( The 3 Killer Advantages )


```python
'''

Hi there 👋 , Now you know what a subagent is …
But the question is :

👉 WHY should you actually use them ?
👉 What do you really gain ?

Here are the 3 KILLER advantages of subagents 🚀


------------------------------------------------------------
(a)  Context Window Isolation
------------------------------------------------------------

👉 Every sub-agent has its OWN context window ..
👉 Main agent's context stays clean & focused ..

Think of it like  ->>   each sub-agent is in its own private room 🚪
   It does its work in its own room ..
   Only the OUTPUT comes out  ->>  not the entire conversation ..


------------------------------------------------------------
(b)  Specialised Sub-agents
------------------------------------------------------------

👉 You can create sub-agents that are SPECIALISED in 1 task ..

For example :
   👉 Auditor Sub-agent   ->>  for code review / audit
   👉 Writer Sub-agent    ->>  for documentation / blogs
   👉 Researcher Sub-agent ->>  for web search / R&D

And the COOL part is :
   👉 You can design every sub-agent as per your OWN need ..

Every agent can have :
   ✅ its own SYSTEM PROMPT
   ✅ its own set of SKILLS
   ✅ its own set of TOOLS  ( only what that agent actually needs )
   ✅ DIFFERENT MODEL access

Example :

   Agent 1   ->>   Opus    ( heavy reasoning  -> use most powerful model )
   Agent 2   ->>   Haiku   ( simple lookup tasks -> use cheaper / faster model )
   Agent 3   ->>   Sonnet  ( balanced tasks  -> mid-tier model )

So you can MIX & MATCH different Claude models for different sub-agents !
This is INSANELY powerful for cost optimization 💰


------------------------------------------------------------
(c)  Parallelism in your Workflow
------------------------------------------------------------

👉 You can run multiple sub-agents IN PARALLEL ..

This is where things get exciting 🚀

Example :
   Imagine you want to perform EDA on 3 different datasets ..

   ❌  WITHOUT sub-agents :
       LLM will do EDA on dataset1  ->>  then dataset2  ->>  then dataset3
       ie  SEQUENTIALLY  ( very slow )

   ✅  WITH sub-agents :
       Spin up 3 sub-agents in parallel
       Each sub-agent does EDA on 1 dataset
       All 3 work AT THE SAME TIME ( parallel )

       Time taken  ->>   roughly 1/3rd of the original time ⚡


Important note :
👉 You can do this ONLY when the tasks are NOT connected to each other ..
   ie  independent tasks ..
   ( If task 2 depends on output of task 1 -> you can't parallelise them )


------------------------------------------------------------
🔹 Quick Recap of Advantages
------------------------------------------------------------

   👉 (a)  Context window isolation       ->>  clean , no overflow
   👉 (b)  Specialised sub-agents          ->>  best agent for the job
   👉 (c)  Parallelism                     ->>  speed + scalability
   👉 BONUS - cost optimization            ->>  mix Opus + Sonnet + Haiku


This is exactly WHY subagents exist in Claude Code ..
and why you will use them in EVERY serious AI project going forward 🔥

'''
```

# .. Top Real-World Use Cases of Subagents


```python
'''

Now lets quickly see :
👉 Where do senior engineers ACTUALLY use sub-agents in real projects ?

Here are the TOP USE-CASES of subagents - super important to know 🔥


------------------------------------------------------------
1)  Codebase Exploration
------------------------------------------------------------

👉 You have a HUGE codebase ( imagine 500 files )
👉 You want Claude to read & understand the entire codebase
👉 If main agent does it  ->>  context window EXPLODES  💥

   Solution -
   👉 Use a sub-agent to crawl the codebase
   👉 It returns only a clean SUMMARY back to main agent ✅


------------------------------------------------------------
2)  Code Review
------------------------------------------------------------

👉 You can use one agent to WRITE the code ..
👉 Then create ANOTHER agent to do CODE REVIEW ..

Why 2 agents ?
👉 Because the agent which wrote the code may NOT do a better code review
   ( same brain , same bias .. )
👉 A separate fresh sub-agent gives you cleaner , unbiased feedback ✅

This is exactly how PRs work in real teams - 
Different person writes , different person reviews 🔥


------------------------------------------------------------
3)  Verification & Validation  /  Testing
------------------------------------------------------------

👉 One sub-agent writes the feature
👉 Another sub-agent writes the test cases / validates the feature
👉 Clean , focused , reliable 🚀


------------------------------------------------------------
4)  Multi-stage Pipeline
------------------------------------------------------------

This is HUGE in data science 👇

   data collection  →  data cleaning  →  EDA   →  modelling

Each stage can be a SEPARATE sub-agent :
   👉 Stage 1 agent  ->  collects data         (I/P  ->  O/P)
   👉 Stage 2 agent  ->  cleans data           (I/P  ->  O/P)
   👉 Stage 3 agent  ->  performs EDA          (I/P  ->  O/P)
   👉 Stage 4 agent  ->  builds the model

Clean pipeline ✅  Clean context ✅  Clean handoff ✅


------------------------------------------------------------
5)  Parallel Independent Tasks
------------------------------------------------------------

👉 As we saw earlier ->>   EDA on 3 datasets in parallel
👉 Or  ->>  testing 3 different APIs at the same time
👉 Or  ->>  running 3 different prompts at once

Anywhere tasks are INDEPENDENT  ->>  use parallel sub-agents ⚡


------------------------------------------------------------
6)  Security Auditing of your code
------------------------------------------------------------

👉 Create a dedicated "Security Auditor" sub-agent
👉 Its only job is to scan code for vulnerabilities ..
👉 SQL injection , API key leaks , unsafe imports etc.

In real production projects , this is GOLD 🪙


------------------------------------------------------------
🔹 Summary
------------------------------------------------------------

   1) Codebase exploration
   2) Code review
   3) Verification & Validation / Testing
   4) Multi-stage pipeline
   5) Parallel independent tasks
   6) Security auditing of your code


Internalise these 6 - because in interviews & real projects ,
these are the EXACT scenarios where you will pull out subagents 🚀

'''
```

# 4..Live-Demo] Types of Subagents  →  
        Built-in Subagents ( Explore , Plan , General-Purpose )


```python
'''

Hi there 👋 , So far we have learnt :
👉 What a sub-agent is
👉 Why we use them


Now lets talk about :
👉 What TYPES of sub-agents do we have in Claude Code ?

In general , Claude Code gives you 2 buckets of sub-agents :

      👉 Built-in  Sub-agents
      👉 Custom    Sub-agents


Lets first deep dive into  ->>  BUILT-IN SUB-AGENTS 🚀


------------------------------------------------------------
🔹 What are Built-in Subagents ?
------------------------------------------------------------

👉 These are sub-agents that Claude Code gives us & we DON'T have to create ..
👉 They get triggered automatically by Claude when needed ..

Claude has  ->>  3 categories of built-in sub-agents :

   1)  Explore  Sub-agent
   2)  Plan     Sub-agent
   3)  General Purpose  Sub-agent


Lets understand each one ✨


------------------------------------------------------------
1)  Explore  Sub-agent
------------------------------------------------------------

👉 Claude will call / invoke / TRIGGER this sub-agent
   when the programmer gives a query like :


    [VS CODE] : 

      "Claude , explore this part of the codebase .."
      "Tell me what this project does .."
      "Summarise this repo .."

Behind the scenes , Claude triggers the EXPLORE sub-agent ..

What does it do ?
   👉 Reads your codebase
   👉 Analyses your codebase
   👉 Creates a clean summary
   👉 Returns the summary back to main agent ✅

Beauty of this :
   👉 Main agent's context window stays UNTOUCHED 🔥
   👉 All the heavy reading happens inside the explore sub-agent's
      isolated context window  ->>  which gets deleted later ..





------------------------------------------------------------
2)  Plan  Sub-agent
------------------------------------------------------------

👉 This sub-agent gets triggered / called by Claude
   when you switch to PLAN MODE in Claude Code & you want to
   generate some IMPLEMENTATION PLAN for some feature ..


    [VS CODE] : 
    
    "Generate an implementation plan for a Sales Analytics
    Dashboard using Python, SQL & Streamlit."
    
     Use : Shift + Tab to switch to Plan Mode

    Claude DOES NOT directly start coding.
    Instead, Plan Sub-agent first thinks :
    
    
           ❓ What is the business goal ?
           ❓ Which KPIs are needed ?
           ❓ Which files/modules should be created ?
           ❓ How should ETL flow work ?
           ❓ How should Streamlit pages be structured ?
    







------------------------------------------------------------
3)  General-Purpose  Sub-agent
------------------------------------------------------------

👉 Whenever Claude feels that he wants a sub-agent for
   READING or WRITING type tasks  ->>  Claude will invoke the General-Purpose sub-agent ..


    Used for :
       👉 Generic file reading
       👉 Generic file writing
       👉 Mixed tasks that don't fit neatly into Explore or Plan


    [VS CODE] :  

    Read the sales dataset and generate a clean business insights report
        summarizing top-selling products, monthly revenue trends,
        and customer purchase behavior
        
    This task is NOT purely :
       ❌ Explore Mode
       ❌ Planning Mode
       
    Because Claude needs to :
       ✅ Read multiple files
       ✅ Process data context
       ✅ Write reports
       ✅ Generate summaries
       ✅ Create documentation
   
   
   So Claude invokes :
       👉 general-purpose sub-agent



------------------------------------------------------------
🔹 Quick Recap : Claude's 3 Built-in Subagents
------------------------------------------------------------

   👉 Explore           ->>  understand codebase  /  summarise repo
   👉 Plan              ->>  generate implementation plans (in Plan mode)
   👉 General-Purpose   ->>  default fallback for read / write tasks


And the BEST part is …
You don't have to "call" any of them manually ..
Claude figures it out for you 🔥


'''
```

# 5.. [Live Demo] : How Sub-agents will be Triggered  ( Automatic  vs  Manual )


```python
'''

Hi there 👋 , Now comes a very common question :

👉 OK , I get it , subagents are powerful ..
   but HOW does Claude actually decide which one to call ?
   Do I need to call them MYSELF every time ?

Great question 🚀  Lets answer it ..


------------------------------------------------------------
🔹 The Good News
------------------------------------------------------------

👉 You DON'T have to manually tell Claude :
      "Hey , go use the Explore sub-agent now .."
      "Hey , go use the Plan sub-agent now .."

Claude is SMART enough to figure this out by itself ..

There are basically 2 ways sub-agents get triggered :

      [diagram]
                          🟦  Sub-agents  🟦
                                |
                  --------------+--------------
                  |                            |
            Automatically                  Manually


Lets break both down 👇


------------------------------------------------------------
1)  Automatic Triggering   ( Default & most common )
------------------------------------------------------------

👉 Claude AUTOMATICALLY figures out  ->>  when to use a sub-agent ..

How ?
   👉 Claude looks at the PROMPT you give ..
   👉 It detects the INTENT  ( explore , plan , write , research etc )
   👉 It picks the RIGHT sub-agent on its own ✅

Examples :

   Prompt  ->> "Explore this codebase & tell me what it does .."
        Claude triggers  →  Explore Sub-agent


   Prompt  ->> "Generate an implementation plan for login feature .."
        Claude triggers  →  Plan Sub-agent


   Prompt  ->> "Read all logs & write me a debug report .."
        Claude triggers  →  General-Purpose Sub-agent


👉 You wrote a NATURAL prompt
👉 Claude understood the intent
👉 Right sub-agent got called  ✨

This is how 90% of real-world Claude Code usage works ..
You don't even REALISE sub-agents are being used behind the scenes 🔥


------------------------------------------------------------
2)  Manual Triggering
------------------------------------------------------------

👉 You can ALSO call a sub-agent MANUALLY ..
👉 ie - YOU explicitly tell Claude  ->>   "Use this sub-agent for this task"

When do we use Manual triggering ?

   👉 When you want FULL CONTROL
   👉 When automatic detection is not picking the right agent
   👉 When you have CUSTOM sub-agents ( we will see them next )
      and you want a SPECIFIC one to run

Example :

   Prompt  ->> "Use the @security-auditor sub-agent to scan this repo .."

   Or even simpler :
   👉 The programmer tells Claude  ->>  "Hey , I need a sub-agent for this task .."


In short  ->>


      Sub-agents
        ↓
   ----------------------
   |                    |
 Automatic            Manually
 (Claude decides       (You tell Claude
  based on prompt)      explicitly)



------------------------------------------------------------
🔹 So which one should YOU prefer ?
------------------------------------------------------------

👉 In 90% of cases  ->>  let it happen AUTOMATICALLY ( its smart enough )
👉 Switch to MANUAL only when -
       - you have built CUSTOM sub-agents
       - you want a very SPECIFIC sub-agent every time
       - you are debugging / fine-tuning workflows



And just to summarise from the previous section :
👉 You can VERIFY which sub-agent was triggered using the observer dashboard ..
👉 So you are never flying blind 👀


In the next section , we will look at  ->>  CUSTOM SUB-AGENTS
( the sub-agents YOU create for your own needs ) 🚀

'''
```

# 9.. Custom Subagents  →  Build Your Own AI Workforce 🛠️


```python
'''

Hi there 👋 , So far we have been talking about BUILT-IN sub-agents
( the ones Claude gives us out of the box ) ..

Now lets unlock the REAL super power :
👉 Building your OWN  ->>  CUSTOM SUB-AGENTS 🔥


------------------------------------------------------------
🔹 What are Custom Sub-agents ?
------------------------------------------------------------

👉 Custom sub-agents are the agents that WE make ourselves
   for our own needs / our own workflows ..

Real-world examples :
   👉 sales-data-analyst  ->>  for our sales pipeline
   👉 jira-ticket-writer  ->>  to write Jira tickets in our company format
   👉 security-auditor    ->>  to scan our APIs for vulnerabilities

These are NOT given by Claude ..
👉 We design them ✅


------------------------------------------------------------
🔹 Types of Custom Sub-agents
------------------------------------------------------------

There are 2 types of Custom Sub-agents :

      [diagram]

                  Custom Sub-agents
                          |
            --------------+--------------
            |                            |
     User-level                    Project-level
     sub-agents                    sub-agents


👉  USER-LEVEL Sub-agents
   - Available at YOUR system level ..
   - You can use them across ALL your projects ✅
   - Stored in your HOME directory  ->>   ~/.claude/

   ie .. once created , every project on your machine can use them 🔥


👉  PROJECT-LEVEL Sub-agents
   - Limited to your CURRENT project ONLY ..
   - They live INSIDE the .claude/ folder of your project
   - ie .. only THIS project can use them

   Path :     <your_project>/.claude/


When to use which one ?
   👉 Common , reusable agents      ->>  USER-LEVEL    ( ~/.claude/ )
   👉 Project-specific agents       ->>  PROJECT-LEVEL ( ./.claude/ )


------------------------------------------------------------
🔹 When you create a Sub-agent , you must configure 6 things
------------------------------------------------------------

These 6 things ->>  define WHAT your sub-agent is & WHAT it can do :


1)  TOOLS ACCESS
   👉 Decide what TOOLS this sub-agent can use ..
   👉 Eg :
       - Explore agent      ->>  MUST have Read tool access
       - Research agent     ->>  WebSearch Tool
       - Coding agent       ->>  Write Tools


2)  SYSTEM PROMPT
   👉 Every sub-agent has its OWN system prompt ..
   👉 This is where YOU explain  ->>  what is the work of this sub-agent
   👉 Think of it as the "job description" of the agent ..


3)  MODELS ACCESS
   👉 Decide which Claude model this sub-agent should use ..
   👉 Eg :
       - Agent 1  ->>  access to Opus models      ( heavy reasoning )
       - Agent 2  ->>  access to Haiku models     ( fast & cheap lookups )
       - Agent 3  ->>  access to Sonnet models    ( balanced )


4)  PERMISSIONS
   👉 You decide what permissions every agent has ..
   👉 Eg : Can it edit files ? Can it run bash ? Can it hit the network ?
   👉 Tighter permissions ->>  safer agents ✅


5)  HOOKS
   👉 Pre / post actions you want around agent runs
   👉 Eg : log every call , run a linter after every write , etc.


6)  SKILLS
   👉 Skills bundles you want this agent to have access to ..
   👉 Eg : "pdf-skill" , "data-analysis-skill" , "frontend-design-skill"



👉 In short  ->>  you can CONFIGURE / CUSTOMIZE all 6 things
   for your sub-agents ✅


------------------------------------------------------------
🔹 Example  ->>  Two Custom Sub-agents
------------------------------------------------------------

      Research Agent                     Code Writer Agent
      ---------------                    ------------------
      Tools  :  Read , Grep ,            Tools  :  Read , Write ,
                WebSearch , etc.                   Grep , Edit
      Model  :  Opus                     Model  :  Sonnet
      Prompt :  "You are a research      Prompt :  "You are a code writer
                 agent .. find , read &              agent .. implement
                 summarise topics .."                 features cleanly .."


See how 2 different agents have :
   👉 DIFFERENT tools
   👉 DIFFERENT models
   👉 DIFFERENT system prompts

Thats the beauty of CUSTOM sub-agents - 
they are TAILORED to the job .


------------------------------------------------------------
🔹 How to CALL these Custom agents ?
------------------------------------------------------------

There are 2 ways :

   👉 1st Way  -  Automatic
        Claude will SEE the agents file & call the RIGHT agent
        based on the prompt you wrote ..

   👉 2nd Way  -  Manual
        You can EXPLICITLY tell Claude :
              "Use the @research-agent for this task .."
              "Use the @code-writer for this part .."

Same concept as built-in agents ✅


------------------------------------------------------------
🔹 Pro Tip - Use the Observer
------------------------------------------------------------

👉 Use the same "agents observe" tool we saw in the demo ..
👉 It works for CUSTOM sub-agents too 🔥
👉 You can see EXACTLY which custom agent is firing & when ..



And with this , you now have the full picture of  ->>  the Subagent system in Claude Code 🎉

In the next section , we will compare  ->>  Skills vs Slash Commands vs Subagents
so you NEVER get confused between them again 🚀

'''
```

# 10.. Skills  vs  Slash Commands  vs  Subagents  →  When to use What ? 🌳


```python
'''

Hi there 👋 , By this point you have learnt :

   👉 Slash Commands     ( from Section 5 )
   👉 Skills              ( from earlier sections )
   👉 Subagents           ( this section )

But heres a confusion almost EVERY learner has :

      "Sir , they all look similar … 
       When should I use which one ?"

GREAT question 🚀 


------------------------------------------------------------
🔹 Quick Definition Refresh
------------------------------------------------------------

👉  SLASH COMMANDS
       Shortcuts you type ( like /explain , /summary )
       to TRIGGER a reusable prompt / workflow ..
       ->>  Save you from typing the same prompt again & again ..


👉  SKILLS
       Reusable folders of instructions + best practices that
       Claude can LOAD when a relevant task comes up ..
       ->>  Encode "how to do X properly" ( docx skill , pdf skill etc )


👉  SUBAGENTS
       Mini AI agents with their OWN isolated context window ,
       OWN tools , OWN model , OWN system prompt ..
       ->>  Used to DELEGATE heavy / parallel / specialised tasks


They look similar - but they live at DIFFERENT layers 👇


------------------------------------------------------------
🔹 The Big Picture
------------------------------------------------------------

      [diagram]

                          🤖  Claude Code  🤖
                                 |
            ---------------------+----------------------
            |                    |                     |
       Slash Commands         Skills                Subagents
      (shortcut prompts)   (how-to playbooks)   (separate AI workers)



------------------------------------------------------------
🔹 Decision Tree  ( SUPER IMPORTANT 🌳 )
------------------------------------------------------------

[decision tree flowchart on screen]


   Q1.   Am I just repeating the SAME prompt over & over again ?
              ↓
        YES  ->>   👉 Use a SLASH COMMAND
                   ( eg  /explain , /summarise , /review )


   Q2.   Do I need Claude to follow a SPECIFIC METHODOLOGY or
         best-practice / format ( like creating a .docx , filling a PDF ,
         building a frontend design ) ?
              ↓
        YES  ->>   👉 Use a SKILL
                   ( Claude loads the right playbook for that file type )


   Q3.   Is the task HEAVY ( codebase exploration , multi-step pipeline ,
         parallel work ) OR does it need a DEDICATED specialist
         ( auditor / writer / researcher ) ?
              ↓
        YES  ->>   👉 Use a SUBAGENT
                   ( own context window , own tools , own model )






------------------------------------------------------------
🔹 Final Takeaway
------------------------------------------------------------

   Repeating a prompt ?           ->>  Slash Command
   Doing a specific kind of file ?->>  Skill
   Need a dedicated AI worker ?   ->>  Subagent



In the next section ,  we will continue building on this foundation
with more powerful Claude Code features ..

See you there 👋

'''
```
